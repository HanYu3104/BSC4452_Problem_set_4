```python
import json
from typing import Any

import requests
import matplotlib.pyplot as plt

SERVICE_URL = 'https://pokeapi.co/api/v2/pokemon/'
HEADERS = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
TIMEOUT_SECONDS = 15


def fetch_pokemon_by_name(
    session: requests.Session,
    pokemon_name: str
) -> dict[str, Any] | None:

    url = SERVICE_URL + pokemon_name.lower()
    print('Retrieving', url)

    try:
        response = session.get(url, timeout=TIMEOUT_SECONDS)
        response.raise_for_status()
        print('Retrieved', len(response.text), 'characters')
        return response.json()

    except requests.exceptions.RequestException as err:
        print('Request error:', err)
        return None

    except ValueError:
        print('Response did not contain valid JSON.')
        return None


def print_pokemon(pokemon: dict[str, Any]) -> None:

    print('\nPokémon name:', pokemon.get('name', 'N/A').title())
    print('Height:', pokemon.get('height', 'N/A'))
    print('Weight:', pokemon.get('weight', 'N/A'))

    print('\nTypes:')
    for pokemon_type in pokemon.get('types', []):
        print('-', pokemon_type['type']['name'].title())

    print('\nAbilities:')
    for ability in pokemon.get('abilities', []):
        print('-', ability['ability']['name'].title())

    print('\nBase stats:')
    for stat in pokemon.get('stats', []):
        stat_name = stat['stat']['name']
        stat_value = stat['base_stat']
        print(f'{stat_name.title()}: {stat_value}')


def graph_pokemon_stats(pokemon: dict[str, Any]) -> None:

    stat_names = []
    stat_values = []

    for stat in pokemon.get('stats', []):
        stat_name = stat['stat']['name'].replace('-', ' ').title()
        stat_value = stat['base_stat']

        stat_names.append(stat_name)
        stat_values.append(stat_value)

    plt.bar(stat_names, stat_values)
    plt.xlabel('Stat')
    plt.ylabel('Base Stat Value')
    plt.title(f"{pokemon.get('name', 'Pokemon').title()} Base Stats")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def main() -> None:

    session = requests.Session()
    session.headers.update(HEADERS)

    while True:
        pokemon_name = input(
            'Enter a Pokémon name to search (blank to quit): '
        ).strip()

        if not pokemon_name:
            break

        pokemon = fetch_pokemon_by_name(session, pokemon_name)

        if not pokemon:
            print('No Pokémon data returned.')
            continue

        print_pokemon(pokemon)
        graph_pokemon_stats(pokemon)


if __name__ == '__main__':
    main()
```
